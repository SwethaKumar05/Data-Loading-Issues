 1. Encoding Issues.
 2. Wrong Delimiter.
 3. Header Misalignment.
 4. Missing or Extra Columns.
 6. Null or Missing Values.
 7.weird or non-standard characters.

 -----


### 1. Encoding Issues.

💥 Problem:
Characters like é, ö, or currency symbols (€, ₹, etc.) appear garbled (e.g. Ã©, â€œ)

---


```python
data = pd.read_csv("", encoding = 'latin1') # Often works for European text
```

```python
data = pd.read_csv("", encoding = 'utf-8')# Default, for general-purpose
```

----

### 2. Wrong Delimiter.

💥 Problem:
The file uses ;, |, or \t (tabs) instead of , as delimiter.


```python
data = pd.read_csv("", delimiter= ';') #or sep = ';'
```

---

### 3. Header Misalignment.

💥 Problem:
First row is not column headers, or extra blank rows at the top.

```python
data = pd.read_csv("", header = 1)
```

---

### 4. Missing or Extra Columns.
💥 Problem:
Columns shift because of quotes or commas inside text fields.

**If the data is wrapped in double quotes ("), treat it as a single field — even if it contains commas.**

Example :  

              1,"Taylor Swift, Ed Sheeran",Pop,150000000

---

```python
data = pd.read_csv("", quotechar = '"')
```

⚠️ Alternative Fix (when rows still fail): error_bad_lines=False (deprecated)
This was used to skip rows that had too many or too few columns:

```python
data = pd.read_csv("", on_bad_lines = 'skip')
```

---

### 5. Null or Missing Values.

💥 Problem:
Some cells are blank or have --, N/A, etc.

**If you see any of these values in a cell, treat them as NaN (missing).**

```python
data = pd.read_csv("", na_values = ['--','N/A', 'null',''])
```

----

### 6. Weird or non-standard characters.

 The Problem
Sometimes a CSV file contains characters that don’t match the encoding you're using to read it — for example:

Smart quotes: “ ” instead of "

Emojis 😃

Special currency symbols: ₹, €, ¥

Characters from other languages (like é, ñ, ğ, etc.)

```python
data = pd.read_csv('file.csv', encoding='utf-8', errors='replace')
```

Use errors='replace' if you want to keep structure but flag unreadable characters.

Use errors='ignore' if you prefer to ignores and removes the problematic characters completely.
